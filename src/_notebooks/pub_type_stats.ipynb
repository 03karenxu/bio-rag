```python
import json
import os
import shutil
from config import DATASET_DIR
from utils.schemas import Paper
from utils.paper_parser import PaperParser
from collections import defaultdict

SET = []
parser = PaperParser()
counts_papers_only_journal = 0
counts_paper_mostly_journal = 0
counts = defaultdict(int)
pmc_lit_reviews = DATASET_DIR / "lit_reviews"
for pmcid_dir in pmc_lit_reviews.iterdir():
    file = next((pmcid_dir / "paper").glob("*.xml"))
    try:
        paper = parser.parse_paper(xml=file)
    except Exception as e:
        print(f"Failed to parse {pmcid_dir}: {e}")
        continue
    
    paper_non_journal = 0
    for ref in paper.references:
        pub_type = ref.pub_type
        counts[pub_type] += 1
        if pub_type != "journal":
            paper_non_journal += 1

    if paper_non_journal == 0:
        counts_papers_only_journal += 1
        SET.append(file)
    elif paper_non_journal <= 5:
        counts_paper_mostly_journal += 1

print(counts)
print(counts_papers_only_journal)
print(counts_paper_mostly_journal)

os.mkdir(DATASET_DIR / "lit_reviews_test")
for file in SET:
    print(file)
    shutil.copytree(file.parent.parent, DATASET_DIR / "lit_reviews_test" / file.parent.parent.name)
```
