```python
from utils.embed import embed_with_retry
import base64
import numpy as np
import litellm

text1 = "dog standing in grass"
text2 = "a boat in the water"
img = "dog.png"

with open(img, "rb") as f:
    base64_image = base64.b64encode(f.read()).decode("utf-8")

# Pick correct MIME type
mime_type = "image/png"

data_url = f"data:{mime_type};base64,{base64_image}"

litellm._turn_on_debug()
embeddings = await embed_with_retry(f"data:image/png;base64,{base64_image}")
```

```python
def cosine_similarity(vec1, vec2):
    vec1 = np.asarray(vec1)
    vec2 = np.asarray(vec2)

    return np.dot(vec1, vec2) / (
        np.linalg.norm(vec1) * np.linalg.norm(vec2)
    )

captions = [
    "dog standing in grass",
    "a photo of a dog standing in green grass outdoors",
    "a brown dog outside on a lawn",
    "a boat in the water"
]

caption_embeddings = await embed_with_retry(to_embed=captions)

image_embedding = embeddings[0]
print("VECTOR SIMILARITIES:")
for i, caption in enumerate(captions):
    print(f"{caption}: ", cosine_similarity(image_embedding, caption_embeddings[i]))
```
